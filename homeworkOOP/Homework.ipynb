Principles of OOP
OOP has four main principles: Encapsulation, Abstraction, Inheritance, and Polymorphism.
A. Encapsulation
* Wrapping data and methods into a single unit (class).
* Protecting data from direct external modification.

```python
class Car:
    def __init__(self):
        self.__max_speed = 200 # Name-mangled attribute
    def get_max_speed(self):
        return self.__max_speed
car = Car()
print(car.get_max_speed()) # 200
```

B. Abstraction
* Hiding complex implementation details.
* Showing only essential features to the user.

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
```

C. Inheritance
* A child class inherits properties and methods from a parent class.
* It promotes code reuse.

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

dog1 = Dog()
dog1.speak()
```

D. Polymorphism
* The same method name can behave differently depending on the object.

```python
class Dog:
    def speak(self):
        print("Dog barks")
class Cat:
    def speak(self):
        print("Cat meows")
animals = [Dog(), Cat()]
for animal in animals:
    animal.speak()
```

```python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price

    def get_price(self):
        return self.__price
    
    def display_info(self):
        print(f"Book title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Price: ${self.__price:.2f}")

class EBook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size

    def get_price(self):
        if self.__price < 0:
            raise ValueError("Price cannot be negative")
        return self.__price
    
    def display_info(self):
        super().display_info()
        print(f"File Size: {self.file_size} MB")

book1 = Book("The Hip Hop Industry", "Mr.GMENGZ", 9.99)
book1.display_info()
book2 = EBook("Kingdom of wonder", "Mrs.Anita", 99.99, 5)
book2.display_info()


    
```
