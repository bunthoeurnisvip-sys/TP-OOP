```python
class Temperature:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (9 / 5) * celsius + 32

    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        return (5 / 9) * (fahrenheit - 32)

    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15

c_temp = 25
f_temp = Temperature.celsius_to_fahrenheit(c_temp)
k_temp = Temperature.celsius_to_kelvin(c_temp)
print(f"Celsius: {c_temp}")
print(f"Fahrenheit: {f_temp}")
print(f"Kelvin: {k_temp}K")
f_temp = 77
c_temp = Temperature.fahrenheit_to_celsius(f_temp)
print(f"Celsius: {c_temp}")
```

```python
class Product:
    store_name = "ITC Online Store"
    product_count = 0

    def __init__(self, name, price, quantity):
        if not self.is_valid_price(price):
            raise ValueError("Price must be positive.")
        self.name = name
        self.price = price
        self.quantity = quantity
        Product.product_count += 1

    def display_product(self):
        print(f"Name: {self.name}")
        print(f"Price: ${self.price}")
        print(f"Quantity: {self.quantity}")
        print(f"Total value: ${self.total_value()}")

    def total_value(self):
        return self.price * self.quantity

    @classmethod
    def get_product_count(cls):
        return cls.product_count

    @staticmethod
    def is_valid_price(price):
        return price > 0

p1 = Product("Laptop", 800, 5)
p2 = Product("Mouse", 15, 20)
p3 = Product("Keyboard", 30, 10)
products = [p1, p2, p3]
print("======================ITC ONLINE STORE=====================")
inventory_total_value = 0
for i in range(len(products)):
    print(f"Product {i + 1}:")
    products[i].display_product()
    inventory_total_value += products[i].total_value()
    print()

print(f"Total number of products: {Product.get_product_count()}")
print(f"Total value of all products: ${inventory_total_value}")
```
